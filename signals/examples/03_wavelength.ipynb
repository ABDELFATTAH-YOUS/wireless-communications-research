```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.widgets import Slider

t = np.linspace(-10, 10, 500)

fig, ax = plt.subplots(figsize=(8, 4))
plt.subplots_adjust(bottom=0.25)

line, = ax.plot(t, np.sin(t))

ax.set_xlim(-10, 10)
ax.set_ylim(-10, 10)
ax.axhline(0) 
ax.axvline(0) 
ax.grid(True)

text = ax.text(
    0.02, 0.95,
    "A = 1\nf = 1",
    transform=ax.transAxes,
    verticalalignment="top"
)

# Slider A
axA = fig.add_axes([0.2, 0.10, 0.6, 0.03])

sliderA = Slider(
    axA, "V",
    1, 10,
    valinit=1,
    valstep=1
)

# Slider f
axF = fig.add_axes([0.2, 0.05, 0.6, 0.03])

sliderF = Slider(
    axF, "f",
    1, 10,
    valinit=1,
    valstep=1
)

def update(value):

    A = sliderA.val
    f = sliderF.val

    y = A * np.sin(2 * np.pi * f * t)

    line.set_ydata(y)

    text.set_text(
        f"A = {A:.1f}\nf = {f:.1f}"
    )

    fig.canvas.draw_idle()

sliderA.on_changed(update)
sliderF.on_changed(update)

plt.show()
```
