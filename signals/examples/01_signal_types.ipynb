#today we will learn how to gnerate signals sin cos noise (random) and also square wave 

```python
import numpy as np
import matplotlib.pyplot as plt 
```

## TIME LINE

```python
t=np.linspace(-2,2,1000) #line time from -2 to 2 with 1000 points
print(t.dtype) #check the type of t
print(t.shape) #check the shape of t
print(f"{t[0:10]} *** {t[-11:-1]}") #print the first 10 values of t
```

## sine and cos

```python


#sine wave

print("Sine wave with frequency 5 ===============================================================")
sine_wave_f5=np.sin(2*np.pi*5*t) #sine wave with frequency 5
print( f"\nsine_wave_f5.dtype: {sine_wave_f5.dtype}") #check the type of sine_wave_f5
print( f"sine_wave_f5.shape: {sine_wave_f5.shape}") #check the shape of sine_wave_f5
print(f"{sine_wave_f5[0:10]} *** {sine_wave_f5[-11:-1]}") #print the first 10 values of sine_wave_f5

print("Sine wave with frequency 1 ===============================================================")
sine_wave_f1=np.sin(2*np.pi*1*t) #sine wave with frequency 1
print( f"\nsine_wave_f1.dtype: {sine_wave_f1.dtype}") #check the type of sine_wave_f1
print( f"sine_wave_f1.shape: {sine_wave_f1.shape}") #check the shape of sine_wave_f1
print(f"{sine_wave_f1[0:10]} *** {sine_wave_f1[-11:-1]}") #print the first 10 values of sine_wave_f1




#cos wave

print("cos wave with frequency 1 ===============================================================")
cos_wave_f1=np.cos(2*np.pi*1*t) #sine wave with frequency 5
print( f"\ncos_wave_f1.dtype: {cos_wave_f1.dtype}") #check the type of cos_wave_f1
print( f"cos_wave_f1.shape: {cos_wave_f1.shape}") #check the shape of cos_wave_f1
print(f"{cos_wave_f1[0:10]} *** {cos_wave_f1[-11:-1]}") #print the first 10 values of cos_wave_f1

print("cos wave with frequency 5 ===============================================================")
cos_wave_f5=np.cos(2*np.pi*5*t) #sine wave with frequency 5
print( f"\ncos_wave_f5.dtype: {cos_wave_f5.dtype}") #check the type of cos_wave_f5
print( f"cos_wave_f5.shape: {cos_wave_f5.shape}") #check the shape of cos_wave_f5
print(f"{cos_wave_f5[0:10]} *** {cos_wave_f5[-11:-1]}") #print the first 10 values of cos_wave_f5
```

```python

fig, (ax1, ax2,ax3,ax4) = plt.subplots(1, 4, figsize=(20, 4))



ax1.spines['top'].set_color('none')
ax1.spines['right'].set_color('none')
ax1.spines['bottom'].set_position(('data', 0))
ax1.spines['left'].set_position(('data', 0))
ax1.set_title("sine wave with frequency 1")
ax1.plot(t, sine_wave_f1, color='purple')
plt.grid(True, linestyle='--')


ax2.spines['top'].set_color('none')
ax2.spines['right'].set_color('none')
ax2.spines['bottom'].set_position(('data', 0))
ax2.spines['left'].set_position(('data', 0))
ax2.set_title("sine wave with frequency 5")
ax2.plot(t, sine_wave_f5, color='purple')

plt.grid(True, linestyle='--')




ax3.spines['top'].set_color('none')
ax3.spines['right'].set_color('none')
ax3.spines['bottom'].set_position(('data', 0))
ax3.spines['left'].set_position(('data', 0))
ax3.set_title("cos wave with frequency 1")
ax3.plot(t, cos_wave_f1, color='purple')
plt.grid(True, linestyle='--')


ax4.spines['top'].set_color('none')
ax4.spines['right'].set_color('none')
ax4.spines['bottom'].set_position(('data', 0))
ax4.spines['left'].set_position(('data', 0))
ax4.set_title("cos wave with frequency 5")
ax4.plot(t, cos_wave_f5, color='purple')

plt.grid(True, linestyle='--')




plt.show()
```

## square wave 

```python
#square wave 

sine=np.sin(2*np.pi*t)
squar_wave=[]
for i in sine:
    if i>=0:
        squar_wave.append(1)
    else:
        squar_wave.append(0)

fig, (ax1) = plt.subplots(1, 1, figsize=(10, 1))



ax1.spines['top'].set_color('none')
ax1.spines['right'].set_color('none')
ax1.spines['bottom'].set_position(('data', 0))
ax1.spines['left'].set_position(('data', 0))
ax1.set_title(" square wave")
ax1.plot(t, squar_wave, color='purple')
plt.grid(True, linestyle='--')

plt.show()

```

## noise wave (random)

```python
#noise wave (random)

noise_wave=np.random.uniform(-1, 1, 1000)
fig, (ax1) = plt.subplots(1, 1, figsize=(10, 1))

ax1.spines['top'].set_color('none')
ax1.spines['right'].set_color('none')
ax1.spines['bottom'].set_position(('data', 0))
ax1.spines['left'].set_position(('data', 0))
ax1.set_title(" noise wave ")
ax1.plot(t, noise_wave, color='purple')
plt.grid(True, linestyle='--')
plt.show()
```
